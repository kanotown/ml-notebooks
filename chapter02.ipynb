## 条件分岐（if文）

**リスト 2.7**　`if`文による条件分岐

```python
score = 85
if score >= 90:
    print("Excellent")
elif score >= 80:
    print("Good")
else:
    print("Bad")
```

### 比較演算子

**リスト 2.8**　比較演算子の使用例

```python
x = 10
y = 7
print(f"x > y  : {x > y}")
print(f"x <= y : {x <= y}")
print(f"x == y : {x == y}")
print(f"x != y : {x != y}")
```

### 論理演算子

**リスト 2.9**　論理演算子による条件の組み合わせ

```python
temp = 25
humidity = 80
is_warm = temp >= 20    # True
is_dry = humidity < 70  # False
print(f"and: {is_warm and is_dry}")
print(f"or : {is_warm or is_dry}")
print(f"not: {not is_warm}")
```

## 繰り返し処理（for文、while文）

### for文

**リスト 2.10**　`for`文によるリストの繰り返し処理

```python
# リストの要素を1つずつ取り出す
fruits = ["りんご", "バナナ", "レモン"]
for fruit in fruits:
    print(f"{fruit}を買います")
```

**リスト 2.11**　`range()`と`if`文の組み合わせ

```python
# 0〜9のうち3の倍数だけを表示する
for i in range(10):
    if i % 3 == 0:
        print(f"{i} は3の倍数")
```

### while文

**リスト 2.12**　`while`文による素数判定

```python
# n (>2) が素数かどうかを判定するプログラム
n = 23
i = 2
while i < n:
    if n % i == 0:
        print(f"{n} は素数ではない")
        break
    i = i + 1
if i == n:
    print(f"{n} は素数")
```

## 関数

**リスト 2.13**　関数の定義と呼び出し

```python
# 足し算を行う関数
def add(a, b):
    return a + b

# 3 の倍数かどうかを判定する関数
def is_3x(n):
    return n % 3 == 0

# リストの平均値を計算する関数
def calc_average(numbers):
    return sum(numbers) / len(numbers)

# 関数の呼び出し
print(add(3, 5))
print(is_3x(42))
print(calc_average([10, 20, 30, 40]))
```

## ライブラリの使い方（NumPy、pandas、Matplotlib）
