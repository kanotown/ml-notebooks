### 複数グラフの描画

**リスト 3.11**　1行2列のグラフ配置

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(100)
y1 = np.random.normal(100, 10, 100)
y2 = np.random.normal(80, 15, 100)
y3 = np.random.normal(60, 12, 100)
y4 = np.random.normal(45, 10, 100)

fig, axs = plt.subplots(1, 2, figsize=(5.8, 2.4), sharey=True)

axs[0].plot(x, y1, label="系列A")
axs[0].plot(x, y2, label="系列B")
axs[0].set_title("データセット1")
axs[0].set_xlabel("インデックス1")
axs[0].set_ylabel("値")
axs[0].legend()

axs[1].plot(x, y3, label="系列C")
axs[1].plot(x, y4, label="系列D")
axs[1].set_title("データセット2")
axs[1].set_xlabel("インデックス2")
axs[1].legend()

plt.tight_layout()
plt.show()
```

**リスト 3.12**　2行2列のグラフ配置

## 探索的データ分析（EDA）

### データの準備

**リスト 3.17**　Irisデータセットの読み込み

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Iris データセットの読み込み（seaborn のサンプルデータ）
df = sns.load_dataset("iris")

print(df.shape)
print(df.head())
```

### データの分布の確認

#### ヒストグラムによる分布の確認

**リスト 3.18**　ヒストグラムによる分布の確認

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(7.2, 4.6))
feature_cols = ["sepal_length", "sepal_width",
                "petal_length", "petal_width"]

for i, ax in enumerate(axes.flatten()):
    sns.histplot(data=df, x=feature_cols[i], hue="species",
                 kde=True, ax=ax)

plt.tight_layout()
plt.show()
```

#### 箱ひげ図による分布の確認

**リスト 3.19**　箱ひげ図による分布の確認

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(6.6, 4.6))
feature_cols = ["sepal_length", "sepal_width",
                "petal_length", "petal_width"]

for i, ax in enumerate(axes.flatten()):
    sns.boxplot(data=df, x="species", y=feature_cols[i], ax=ax)

plt.tight_layout()
plt.show()
```

### 外れ値の検出

#### 箱ひげ図による外れ値の確認

**リスト 3.20**　箱ひげ図による外れ値の確認

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(5.5, 2.8))
sns.boxplot(data=df[feature_cols])
plt.title("各特徴量の箱ひげ図")
plt.ylabel("値 (cm)")
plt.tight_layout()
plt.show()
```

#### IQR法による外れ値の検出

**リスト 3.21**　IQR法による外れ値の検出

```python
# IQR法で外れ値を検出する
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower) | (df[column] > upper)]
    return outliers

# sepal width の外れ値を検出
col = "sepal_width"
outliers = detect_outliers_iqr(df, col)
print(f"{col} の外れ値の数: {len(outliers)}")
print(outliers[[col, "species"]].head())
```

### 相関分析

#### 相関行列の計算

**リスト 3.22**　相関行列の計算

```python
# 数値列のみで相関行列を計算
corr_matrix = df[feature_cols].corr()
print(corr_matrix.round(2))
```

#### ヒートマップによる相関の可視化

**リスト 3.23**　相関行列のヒートマップ表示

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6.3, 4.7))
sns.heatmap(corr_matrix, annot=True,
            fmt=".2f", cmap="coolwarm",
            center=0, square=True,
            linewidths=0.5)
plt.title("特徴量間の相関行列")
plt.tight_layout()
plt.show()
```
